```python
import pandas as pd
import numpy as np
from datetime import datetime
date_range=pd.date_range(start='1-1-2022',end='31-12-2022')
data=pd.DataFrame(date_range,columns=['date'])
data['data']=(np.random.randint(1,20,size=(len(date_range))))
data.set_index('date',inplace=True)
```

```python
data
```

```python
#resample('M'): group data by month (assumes DateTimeIndex).
#.mean(): take average for each month.

df=data.resample('M').mean()
df
```

```python
import pandas as pd
import numpy as np
# we need to make a dataframe with multiindex
index=pd.MultiIndex.from_arrays([np.array(['A','B','C']),np.array(['one','two','three'])],names=['category','subcategory']) # index olumns are not counted.
data=pd.DataFrame(np.random.randint(1,20,size=(3,2)),index=index,columns=['value1','value2'])
```

```python
print(data.loc['A']) #iloc uses integer position not strings or characters
# indexing at index 'A'
print(data.loc['B','two'])
```

```python
# calculating the sum of each category and sub category.
data.groupby('category')[['value1','value2']].agg(['sum'])
```
