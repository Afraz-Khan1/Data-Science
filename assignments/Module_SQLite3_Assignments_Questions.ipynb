```python
import sqlite3
connection=sqlite3.connect('test4.db')
cursor=connection.cursor()

cursor.executescript('''
  pragma foreign_keys= off;
  begin transaction;
  alter table students rename to old_students;
  create table new_student(
      id integer primary key,
      name text not null,
      age integer,
      department text,
      department_id integer,
      foreign key (department_id) references department(id)
    );
    insert into new_student(id,name,age,department)
    select id,name,age,department from old_students;
    drop table old_students;
    commit;
    pragma foreign_keys=on;
''')

#BEGIN TRANSACTION; starts a manual transaction — it lets you group multiple SQL operations together.

#You control when to:

#COMMIT; → Save all changes.

#ROLLBACK; → Undo all changes since BEGIN.

#Useful for ensuring atomic operations (all succeed or none).
```

```python
connection=sqlite3.connect('test4.db')
cursor=connection.cursor()
def adding_data(department_id,department_name,student_id,student_name,student_age):
    try:
     cursor.execute('''
   insert into department(id,name)
                   values(?,?)
   ''',(department_id,department_name))

     cursor.execute('''
     insert into new_student(id,name,age,department,department_id)
               values(?,?,?,?,?)
     ''',(student_id,student_name,student_age,department_name,department_id))
     connection.commit()
    except Exception as e:
        print("an error occured while inserting.")
        print(e)
    finally:
      connection.close()
adding_data(234,'csit',21,'fraz',20) # after running it one time connection is now closed of the database in finally.
```
