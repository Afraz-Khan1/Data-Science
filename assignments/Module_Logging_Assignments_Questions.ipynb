```python
import logging
def logg():
   logging.basicConfig(
    level=logging.DEBUG,
    format="%(asctime)s %(name)s %(levelname)s %(message)s",
    datefmt='%Y-%m-%d %H:%M:%S',
    handlers=[
       logging.FileHandler("app1.log"), # manage data in file.
     ]
    )
   logging.debug("this is a debug message")
logg()
```

```python
import logging
def logg():
   logging.basicConfig(
    level=logging.DEBUG,
    format="%(asctime)s %(name)s %(levelname)s %(message)s",
    datefmt='%Y-%m-%d %H:%M:%S',
    handlers=[
       logging.FileHandler("app.log"),
       logging.StreamHandler() # show messages in console here is not included message not shown here only in file.
     ]
    )
   logger1=logging.getLogger("module1")
   logger1.setLevel(logging.WARNING)
   logger1.warning("this is warning message")
   logger1.debug("this is a debug message") # as level is warning so debug will not be shown but all other below warning will be shown. 
   logger1.info('This is an info message') # this level is also higher than warning also so not shown either
   logger1.error('This is an error message')
   logger1.critical('This is a critical message')

logg()
```

```python
# also add some contextual keywords in format.
import logging
def logg():
    logging.basicConfig(
    level=logging.DEBUG,
    format="%(asctime)s %(name)s %(levelname)s %(fileName)s %(lineno)d %(message)s",
    datefmt='%Y-%m-%d %H:%M:%S',
    handlers=[
       logging.FileHandler("app.log"),
       logging.StreamHandler() # show messages in console here is not included message not shown here only in file.
     ]
    )
    def fun():
       logging.debug("this is a debug message")
       logging.warning("this is a warning message")
       logging.info("this is info message")
       logging.exception("this is a exception message")
       logging.critical("this is a critical message")
       logging.error("this is a error message")
    fun()
logg()
```
