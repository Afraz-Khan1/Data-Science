### Polynomial regression

```python
#where the datapoints are non linear and following a u shape. so we need to find best fit line i ushape
# polynomial have degrees like 0
#ho(x)=y^=bo x^0
#degree 1
#ho(x)=y^=bo x^0 +b1 x^1   <-same as simple linear regression eq
#degree 2
#ho(x)=y^=bo x^0 +b1 x^1+ b2 x^2
#so on.
```

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
```

```python
x = 6 * np.random.rand(100, 1) - 3
y =0.5 * x**2 + 1.5*x + 2 + np.random.randn(100, 1)
# quadratic equation used- y=0.5x^2+1.5x+2+outliers
plt.scatter(x,y,color='g')
plt.xlabel('x dataset')
plt.ylabel('y dataset')
```

```python
from sklearn.model_selection import train_test_split
xtrain,xtest,ytrain,ytest=train_test_split(x,y,train_size=0.2,random_state=42)
```

```python
from sklearn.linear_model import LinearRegression
regression=LinearRegression()
```

```python
regression.fit(xtrain,ytrain)
```

```python
ypred=regression.predict(xtest)
```

```python
from sklearn.metrics import r2_score
score=r2_score(ytest,ypred)
print(score)
```

```python
plt.plot(xtrain,regression.predict(xtrain),color='r')
plt.scatter(xtrain,ytrain)
plt.xlabel("X Dataset")
plt.ylabel("Y")
```

```python
from sklearn.preprocessing import PolynomialFeatures
poly=PolynomialFeatures(degree=2,include_bias=True)
xtrainpoly=poly.fit_transform(xtrain)
xtestpoly=poly.transform(xtest)
```

```python
xtrainpoly
```

```python
xtestpoly
```

```python
from sklearn.linear_model import LinearRegression
regression=LinearRegression()
regression.fit(xtrainpoly,ytrain)
pred=regression.predict(xtestpoly)
score=r2_score(ytest,pred)
print(score)
```

```python
plt.scatter(xtrain,regression.predict(xtrainpoly))#best fit line
plt.scatter(xtrain,ytrain,color='r')
```

```python
#3 Prediction of new data set
X_new = np.linspace(-3, 3, 200).reshape(200, 1)
X_new_poly = poly.transform(X_new)
X_new_poly
```

```python
y_new = regression.predict(X_new_poly)
plt.plot(X_new, y_new, "r-", linewidth=2, label=" New Predictions")
plt.plot(xtrain, ytrain, "b.",label='Training points')
plt.plot(xtest, ytest, "g.",label='Testing points')
plt.xlabel("X")
plt.ylabel("y")
plt.legend()
plt.show()
```

## Pipeline Concepts 

```python
from sklearn.pipeline import Pipeline
```

```python

def poly_regression(degree):
      xnew = np.linspace(-3, 3, 200).reshape(200, 1) #creating a standardized dataset
      poly_features=PolynomialFeatures(degree=degree,include_bias=True)
      regression1=LinearRegression()
      poly_regression=Pipeline([
        ("poly_features",poly_features),("regression1",regression1)
      ]) 
      poly_regression.fit(xtrain,ytrain)
      ypred=poly_regression.predict(xnew)
      plt.plot(xnew,ypred,"r.",label='degree' + str(degree),linewidth=2)
      plt.plot(xtrain,ytrain,"b.",linewidth=3)
      plt.plot(xtest,ytest,"g.",linewidth=3)
      plt.legend(loc="upper left")
      plt.xlabel("X")
      plt.ylabel("y")
      plt.axis([-4,4,0,10])
      plt.show()
```

```python
poly_regression(4)
```
