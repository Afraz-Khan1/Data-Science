```python
import seaborn as sns
df=sns.load_dataset('tips')
```

```python
df.head()
```

```python
df['time'].value_counts
```

```python
from sklearn.preprocessing import OneHotEncoder,LabelEncoder
lab=LabelEncoder()
hot_for_size=OneHotEncoder()
hot_for_day=OneHotEncoder() # we can't use same onehot encoder for both day and size as one got trained for only one column.
df['sex']=lab.fit_transform(df['sex'])
df['smoker']=lab.fit_transform(df['smoker'])
df['time']=lab.fit_transform(df['time'])
day=hot_for_day.fit_transform(df[['day']]).toarray() # now we have sparse matrix in the day 
size=hot_for_size.fit_transform(df[['size']]).toarray() # and also size is the sparse matrix of the size.
```

```python
encoded_day=pd.DataFrame(day,columns=hot_for_day.get_feature_names_out())
encoded_size=pd.DataFrame(size,columns=hot_for_size.get_feature_names_out())
```

```python
encoded_day
```

```python
encoded_size
```

```python
df=pd.concat([df,encoded_day,encoded_size],axis=True)
```

```python
df.drop(['day','size'],axis=1,inplace=True)
```

```python
df.head()
```

```python
df.columns
```

```python
x=df.drop('time',axis=1) # only return columns except time to store in x
y=df['time']
```

```python
from sklearn.model_selection import train_test_split
xtrain,xtest,ytrain,ytest=train_test_split(x,y,test_size=0.3,random_state=0)
```

```python
from sklearn.naive_bayes import BernoulliNB
model=BernoulliNB()
model.fit(xtrain,ytrain)
```

```python
ypred=model.predict(xtest)
```

```python
from sklearn.metrics import confusion_matrix,accuracy_score,classification_report
print(confusion_matrix(ytest,ypred))
print(accuracy_score(ytest,ypred))
print(classification_report(ytest,ypred))
```

```python
params = {
    'alpha': [0.001, 0.01, 0.1, 1.0, 10.0],
    'fit_prior': [True, False],
    'binarize': [0.0, 0.5]
}
```

```python
#hyper tunning
from sklearn.model_selection import GridSearchCV
grid=GridSearchCV(estimator=model,param_grid=params,scoring='accuracy',cv=5,n_jobs=-1)
```

```python
grid.fit(xtrain,ytrain)
```

```python
ypred=grid.predict(xtest)
```

```python
from sklearn.metrics import confusion_matrix,accuracy_score,classification_report
print(confusion_matrix(ytest,ypred))
print(accuracy_score(ytest,ypred))
print(classification_report(ytest,ypred))
```

```python
# First check the target column (y) to decide the type of problem.

# 1. If y is CONTINUOUS (numbers with many possible values)
# Example: price, salary, temperature
# Use REGRESSION models:
# - Linear Regression
# - Ridge/Lasso Regression
# - Random Forest Regressor
# - SVR


# 2. If y is CATEGORICAL (classes)
# Example: 0/1, Yes/No, Cat/Dog
# Use CLASSIFICATION models:
# - Logistic Regression
# - SVM
# - Decision Tree
# - Random Forest
# - Naive Bayes


# For Naive Bayes specifically:

# Gaussian Naive Bayes:
# - y must be categorical (binary or multiclass)
# - X can contain continuous numerical features
# Example:
# X = age, salary, height
# y = disease yes/no
# Use GaussianNB


# Bernoulli Naive Bayes:
# - y must be categorical (binary or multiclass)
# - X features should be binary (0/1)
# Example:
# X = smoker(0/1), gender(0/1), word_present(0/1)
# y = spam/not spam
# Use BernoulliNB


# Multinomial Naive Bayes:
# - y must be categorical
# - X contains counts/frequencies
# Example:
# X = word counts in documents
# y = category of document
# Use MultinomialNB
```
