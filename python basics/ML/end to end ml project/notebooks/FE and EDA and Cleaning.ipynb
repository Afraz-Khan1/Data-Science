## Algerian Forest Fires Dataset 
Data Set Information:

The dataset includes 244 instances that regroup a data of two regions of Algeria,namely the Bejaia region located in the northeast of Algeria and the Sidi Bel-abbes region located in the northwest of Algeria.

122 instances for each region.

The period from June 2012 to September 2012.
The dataset includes 11 attribues and 1 output attribue (class)
The 244 instances have been classified into fire(138 classes) and not fire (106 classes) classes.

Attribute Information:

1. Date : (DD/MM/YYYY) Day, month ('june' to 'september'), year (2012)
Weather data observations
2. Temp : temperature noon (temperature max) in Celsius degrees: 22 to 42
3. RH : Relative Humidity in %: 21 to 90
4. Ws :Wind speed in km/h: 6 to 29
5. Rain: total day in mm: 0 to 16.8
FWI Components
6. Fine Fuel Moisture Code (FFMC) index from the FWI system: 28.6 to 92.5
7. Duff Moisture Code (DMC) index from the FWI system: 1.1 to 65.9
8. Drought Code (DC) index from the FWI system: 7 to 220.4
9. Initial Spread Index (ISI) index from the FWI system: 0 to 18.5
10. Buildup Index (BUI) index from the FWI system: 1.1 to 68
11. Fire Weather Index (FWI) Index: 0 to 31.1
12. Classes: two classes, namely Fire and not Fire

## DATA CLEANING

```python
#axis=1 → check across columns (row-wise)
#.any() → returns True if any value in that row is True (i.e., at least one missing value exists).
df[df.isnull().any(axis=1)] # region has all blanks next to it as it is header and header has all column blank.
```

As the dataset is divided into two regions. when you see whole dataset at once you see divided in two sets.
The dataset is converted into two sets based on Region from 122th index, we can make a new column based on the Region

1 : "Bejaia Region Dataset"

2 : "Sidi-Bel Abbes Region Dataset"

Add new column with region

```python
df.loc[:122,"Region"]=0
df.loc[122:,"Region"]=1
```

```python
df.info()
```

```python
df["Region"]=df["Region"].astype(int)
```

```python
df.info()
```

```python
df.isnull().sum()
```

```python
df=df.dropna().reset_index(drop=True)
```

```python
df.isnull().sum()
```

```python
df.iloc[[122]] # this index has redundant column names we emit it out. as new column region is already added for two different regions.
```

```python
# remove the 122 nd row.
df=df.drop(122).reset_index(drop=True)
```

```python
df.iloc[[122]]
```

```python
df.columns
```

```python
# remove extra spaces from the column names in order to make column name more accessable.
df.columns=df.columns.str.strip()
df.columns
```

 Change the required column data type as integer

```python
df[['day','month','year','Temperature','RH','Ws']]=df[['day','month','year','Temperature','RH','Ws']].astype(int)
```

```python
df.info()
```

```python
df.columns
```

```python
# but in some cases freature can be in quantity then we should use a general approch to convert type in bulk.
objects=[features for features in df.columns if df[features].dtypes=='O'] #0 means object type.

for i in objects:
    if i!='Classes':
       df[i]=df[i].astype(float)
```

```python
#replacing fire and not fire with fire and not fire to recognize them as a whole in df where we didn't changed the classes yet.
df['Classes']=np.where(df['Classes'].str.contains('not fire'),'not fire','fire')
```

```python
#ploting a subplot graph( having multiple plots in a single graph). using df and its feature months with hue(which seperates each plot in category
# of different colours.) using hue=classes in df.


plt.subplots(figsize=(13,6))
sns.set_style('whitegrid')
sns.countplot(x='month',hue='Classes',data=df)
plt.xlabel('Months',weight='bold')
plt.ylabel('Classes',weight='bold')
plt.title('Fire analysis of both region')
plt.show()
```

Its observed that August and September had the most number of forest fires for both regions. And from the above plot of months, we can understand few things
