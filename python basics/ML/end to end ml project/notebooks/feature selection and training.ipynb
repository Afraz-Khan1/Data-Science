```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
df=pd.read_csv('Algerian_forest_fires_cleaned_dataset.CSV')
```

```python
#Predict Classes → Classification problem→ FWI is a feature.
#Predict FWI(continuous) → Regression problem → FWI is the target. all other are features.
# if we want to predict classes( as if we keep classes as our target) than it is a classification problem as classes has fire not fire category.
# if we want to predict fwi(fire weather index our target) it is continuous so it is our regression problem. and we keep classes as our independent feature.
```

```python
# removing day month year as we dont need it currently.
df.drop(['day','month','year'],axis=1,inplace=True)
```

```python
df.head()
```

```python
#encoding classes
df['Classes']=np.where(df['Classes'].str.contains('not fire'),0,1)
```

```python
df['Classes'].value_counts() # fixed
```

```python
#dividing features into depended and independent
x=df.drop(['FWI'],axis=1) # x has all the independent feature
y=df['FWI'] # y has the dependent feature that is predicted by all other independent features.
```

```python
# train test split
from sklearn.model_selection import train_test_split
xtrain,xtest,ytrain,ytest=train_test_split(x,y,test_size=0.25,random_state=42)
```

```python
xtrain.shape,xtest.shape
```

 In feature selection
 Example of highly correlated features:
 Temperature and Heat Index may have a very high correlation.
 Both features provide similar information about heat conditions.

 Keeping both features can cause:
 1. Multicollinearity (model gets confused about feature importance)
 2. Unstable coefficients in models like Linear Regression
 3. Extra unnecessary features that increase complexity

 Therefore, we remove one of the highly correlated features
 and keep the most useful feature for the model.

```python
#feature selection by checking the correlation if it is worth it to keep a feature or it has not a significant affect on the target prediction.
```

```python
# check for multicollinearity
plt.figure(figsize=(12,10))
sns.heatmap(xtrain.corr(),annot=True)
```

## Feature Scaling Or Standardization

```python
# find the feature that are highly correlated than remove them.
def correlation(df,thershold):
    high_corr_cols=set() #made empty set in order to handle uniqueness
    correlation_matrix=df.corr() #df.corr() actually calculates the corr for every column and have a whole matrix of it.
    for i in range(len(correlation_matrix.columns)):
        for j in range(i):
          if abs(correlation_matrix.iloc[i,j]) > thershold:
             column=correlation_matrix.columns[i]
             high_corr_cols.add(column)
    return high_corr_cols

# i = current feature (current row/column)
# j = previous feature to compare with

# correlation_matrix.iloc[i, j]
# -> Gets the correlation between feature i and feature j.

# correlation_matrix.columns[i]
# -> Gets the name of feature i.

# If feature i is highly correlated with feature j,
# we keep feature j and remove feature i.
#
# Example:
# A and B have correlation = 0.95
# Keep A, remove B.
# So: column = correlation_matrix.columns[i]  # gets 'B'
```

```python
corre_features=correlation(xtrain,0.85)
```

```python
xtrain.drop(corre_features,axis=1,inplace=True)
xtest.drop(corre_features,axis=1,inplace=True)
xtrain.shape,xtest.shape
```

# standardization

```python

from sklearn.preprocessing import StandardScaler

scalar=StandardScaler()
xtrain_scaled=scalar.fit_transform(xtrain)
xtest_scaled=scalar.transform(xtest)
```

```python
xtrain_scaled
```

```python
plt.figure(figsize=(15,5))
plt.subplot(1,2,1)
sns.boxplot(xtrain)
plt.title('xtrain before scaling')
plt.subplot(1,2,2)
sns.boxplot(xtrain_scaled)
plt.title('xtrain after scaling')

# Split the figure into 1 row and 2 columns, and select the 2nd plot.
#plt.subplot(1,2,2)
```

# Linear Regression

```python
from sklearn.linear_model import LinearRegression
from sklearn.metrics import mean_absolute_error
from sklearn.metrics import r2_score
regression=LinearRegression()
regression.fit(xtrain_scaled,ytrain)
ypred=regression.predict(xtest_scaled)
mae=mean_absolute_error(ytest,ypred) #yest actual value ypred predicted value.
score=r2_score(ytest,ypred)
print("Mean absolute error", mae)
print("R2 Score", score)
plt.scatter(ytest,ypred)
```

# Lasso Regression

```python
# every thing is same just replaced regression with lasso.
# used for feature selection is on target not useless features are used.
from sklearn.linear_model import Lasso
from sklearn.metrics import mean_absolute_error
from sklearn.metrics import r2_score
lasso=Lasso()
lasso.fit(xtrain_scaled,ytrain)
y_pred=lasso.predict(xtest_scaled)
mae=mean_absolute_error(ytest,y_pred)
score=r2_score(ytest,y_pred)
print("Mean absolute error", mae)
print("R2 Score", score)
plt.scatter(ytest,y_pred)
```

#Cross validation Lasso Regression

```python
from sklearn.linear_model import LassoCV
lassocv=LassoCV(cv=5) # cv =5 means it will have 5 folds of the mse
lassocv.fit(xtrain_scaled,ytrain)
```

```python
lassocv.alpha_ #alpha value.
```

```python
lassocv.alphas_ # total 100 alphas parameter tested on the dataset
```

```python
lassocv.mse_path_ # as there are 5 folds in cv mse and 100 alphas. there are 500 mse path values.
# lassoCV.mse_path_ stores the MSE (Mean Squared Error) for every alpha value
# tested during cross-validation.

# Example:
# 100 alphas × 5 folds = 500 MSE values

# mse_path_ tells us:
# -> How well the model performs for each alpha.
# -> Lower MSE means better performance.

# alpha tells us:
# -> Strength of regularization.
# -> Small alpha = less penalty (more features kept)
# -> Large alpha = more penalty (more features shrink to zero)

# LassoCV chooses the alpha with the lowest average MSE.
```

```python
# after doing croos validation.
y_pred=lassocv.predict(xtest_scaled)
plt.scatter(ytest,y_pred)
mae=mean_absolute_error(ytest,y_pred)
score=r2_score(ytest,y_pred)
print("Mean absolute error", mae)
print("R2 Score", score)
```

# Ridge Regression

```python
# to reduce overfitting
from sklearn.linear_model import Ridge
from sklearn.metrics import mean_absolute_error
from sklearn.metrics import r2_score
ridge=Ridge()
ridge.fit(xtrain_scaled,ytrain)
ypred=ridge.predict(xtest_scaled)
mae=mean_absolute_error(ytest,ypred)
score=r2_score(ytest,ypred)
print("Mean absolute error", mae)
print("R2 Score", score)
plt.scatter(ytest,ypred)
```

#Cross validation ridge Regression

```python
from sklearn.linear_model import RidgeCV
ridgecv=RidgeCV(cv=5)
ridgecv.fit(xtrain_scaled,ytrain)
```

```python
ypred=ridgecv.predict(xtest_scaled)
plt.scatter(ytest,ypred)
score=r2_score(ytest,ypred)
mae=mean_absolute_error(ytest,ypred)
print("mean absolute error",mae)
print("r2 score",score)
```

# Elastic net regression
