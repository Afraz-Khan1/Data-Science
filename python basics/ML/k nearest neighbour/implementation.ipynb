```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
from sklearn.datasets import make_classification

x,y=make_classification(n_samples=1000,n_features=3,n_redundant=1,random_state=999,n_classes=2)
# so we used here binary classification.
# n_redundant is the feature that can be obtained by the information combination of other column but its kept seperate so that column is redundant

# Without random_state:
# Run 1 → Train/Test split A → accuracy 90%
# Run 2 → Different split B → accuracy 85%
# Run 3 → Different split C → accuracy 92%
#
# You don't know whether the accuracy changed because of:
# 1. The model
# 2. The data split
#
# With random_state=42:
# Same Train/Test split every run → easier to compare models fairly
# and reproduce the results.
#
# Important:
# random_state does NOT make training and testing data the same.
# It only keeps the SAME SPLIT every time. [1 ,2 ,3 ,4] train [5, 6, 7] test. same data every run .
```

```python
from sklearn.model_selection import train_test_split
xtrain,xtest,ytrain,ytest=train_test_split(x,y,test_size=0.33,random_state=42)
```

```python
from sklearn.neighbors import KNeighborsClassifier
classifier=KNeighborsClassifier(n_neighbors=5,algorithm='auto')
classifier.fit(xtrain,ytrain)
ypred=classifier.predict(xtest)
```

```python
from sklearn.metrics import confusion_matrix,accuracy_score,classification_report
print(confusion_matrix(ytest,ypred))
print(accuracy_score(ytest,ypred))
print(classification_report(ytest,ypred))
```

```python
params={
    'n_neighbors':range(1,11)
}
from sklearn.model_selection import GridSearchCV
grid=GridSearchCV(estimator=KNeighborsClassifier(),param_grid=params,scoring='accuracy',cv=5)
```

```python
grid.fit(xtrain,ytrain)
ypred=grid.predict(xtest)
```

```python
print(confusion_matrix(ytest,ypred))
print(accuracy_score(ytest,ypred))
print(classification_report(ytest,ypred))
```
