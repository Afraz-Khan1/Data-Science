# now starting feature engineering column wise

```python
df['day']=df['Date_of_Journey'].str.split('/').str[0]  #so first str because object type column is actually string. and last str to pick the first splited value
df['month']=df['Date_of_Journey'].str.split('/').str[1]
df['year']=df['Date_of_Journey'].str.split('/').str[2]
```

```python
df.info() # new columns added with seperate day month year value
```

```python
# but inorder to make these column ML model understandable we need to encode them but for that they should be of type int
df['day']=df['day'].astype(int)
df['month']=df['month'].astype(int)
df['year']=df['year'].astype(int)
```

```python
df.info()
```

```python
df.drop('Date_of_Journey',axis=1,inplace=True)
```

```python
df.head()
```

```python
# so we need to remove the extra month written in arrival time and extract the time hour and minute
df['Arrival_Time']=df['Arrival_Time'].apply(lambda x:x.split(' ')[0])
```

```python
df.head()
```

```python
df['Arrival_hour']=df['Arrival_Time'].str.split(':').str[0]
df['Arrival_min']=df['Arrival_Time'].str.split(':').str[1]
```

```python
df.drop('Arrival_Time',axis=1,inplace=True)
```

```python
df.head()
```

```python
df['Arrival_hour']=df['Arrival_hour'].astype(int)
df['Arrival_min']=df['Arrival_min'].astype(int)
```

```python
df.info()
```

```python
df['Departure_hour']=df['Dep_Time'].str.split(':').str[0]
df['Departure_min']=df['Dep_Time'].str.split(':').str[1]
```

```python
df.info()
```

```python
df['Departure_hour']=df['Departure_hour'].astype(int)
df['Departure_min']=df['Departure_min'].astype(int)
```

```python
df.drop('Dep_Time',axis=1,inplace=True)
```

```python
# converting departure time into hour floats
# so when the duration is written in this format 2h 50m we use this r'(\d+)h'. r tells raw string. () tells to capture what ever is returned. \d finds
# the number written like from 2h = 2. and h tells to find only that number written with h.

hour=df['Duration'].str.extract(r'(\d+)h').fillna(0).astype(float)  # fillna fills 0 where there is no hour written. so hour will have 
min=df['Duration'].str.extract(r'(\d+)m').fillna(0).astype(float)

df['Duration_hour']=hour + min/60
```

```python
df.head()
```

```python
df.drop(columns='Duration',axis=1,inplace=True)
```

```python
df.head()
```

```python
# dropping route as we have source and destination already in the dataset
df.drop(columns='Route',axis=1,inplace=True)
```

```python
#now we need to convert all categorical features into numerical values using one ht encoding
from sklearn.preprocessing import OneHotEncoder
encoder=OneHotEncoder()
```

```python
encoder.fit_transform(df[['Destination','Source','Airline']]).toarray()
```

```python
encoded_df=pd.DataFrame(encoder.fit_transform(df[['Destination','Source','Airline']]).toarray(),columns=encoder.get_feature_names_out())
```

```python
#now drop the other columns and merge these encoded into datasets.
df=pd.concat([df,encoded_df],axis=1)
```

```python
df.drop(columns=['Source','Airline','Destination'],axis=1,inplace=True)
```

```python
df.head()
```

```python
df['Total_Stops'].unique()
```

```python
#odinal encoding
df['Total_Stops']=df['Total_Stops'].map({'non-stop':0,'1 stop':1,'2 stops':2,'3 stops':3,'4 stops':4,np.nan:1})
```

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
df_play=pd.read_csv('data/google_cleaned.csv')
```

```python
df_play.head()
```

```python
df_play['App'].duplicated().sum()
```

```python
df_play=df_play.drop_duplicates(subset=['App'],keep='first')# keep first only keeps first original and removes copies
```

```python
df_play['App'].duplicated().sum()
```

```python
df_play.shape
```

```python
df_copy=df_play
df_copy.drop(columns='Unnamed: 0',axis=1,inplace=True)
```

```python
df_copy['Last Updated']=df_copy['Last Updated'].astype('datetime64[ns]')
```

```python
numeric_features = [feature for feature in df_copy.columns if df_copy[feature].dtype != 'O']
categorical_features = [feature for feature in df_copy.columns if df_copy[feature].dtype == 'O']

# print columns
print('We have {} numerical features : {}'.format(len(numeric_features), numeric_features))
print('\nWe have {} categorical features : {}'.format(len(categorical_features), categorical_features))
```

```python
# what are the top 10 app category?
plt.figure(figsize=(15, 15))
plt.suptitle('Univariate Analysis of Numerical Features', fontsize=20, fontweight='bold', alpha=0.8, y=1.)

for i in range(0, len(numeric_features)):
    plt.subplot(5, 3, i+1) # adjust current figure to i+1 pos
    sns.kdeplot(x=df_play[numeric_features[i]],fill=True, color='r') # kde kernel density plot area under the curve will be shown throgh fill red.
    plt.xlabel(numeric_features[i])
    plt.tight_layout()
```
