```python
#one hot encoding coverts categorical data into numerical values which represent each category. but its disadvantage is that
# it creates many new columns if there are many features for a single category.
# like colour category has many features like different colours. 
# the matrix that will be created when features are converted into numerical values is caled sparse matrix.
# like when data is seen of 0,0,1 like of three columns together it will be like a matrix so that matrix like looking is called sparse matrix
```

```python
from sklearn.preprocessing import OneHotEncoder
import pandas as pd
#create a colour category with multiple features(shades)
df=pd.DataFrame({
    'colours':['red','green','yellow','red','blue','green']
})
```

```python
#create an instance for the one hot encoder
# than we fit and transform
encoder=OneHotEncoder()
```

```python
df
```

```python
encoded=encoder.fit_transform(df[['colours']]).toarray() # gives you the sparse matrix the matrix that shows numerical values.
```

```python
# convert it in dataframe
encoded_df=pd.DataFrame(encoded,columns=encoder.get_feature_names_out())
```

```python
encoded_df
```

```python
pd.concat([df,encoded_df],axis=1) # as you can see the category numerical values are ready that can be shared with the model.
```

```python
# if you have new data.
encoder.transform([['blue']]).toarray()
```

```python
#label and ordinal encoding
# labeling means to label to the data and ordinal encoding means to rank those labels
# as it is possible that the machine learning model will understand by seeing that the repetition of an feature will be the large one. but we need to stop
# this by ranking.
```

```python
from sklearn.preprocessing import LabelEncoder
df=pd.DataFrame({
    'colours':['red','green','yellow','red','blue','green']
})

```

```python
encoder=LabelEncoder()
```

```python
label=encoder.fit_transform(df[['colours']])
```

```python
label # as model will consider 3 label as the laargest but that can not be true. so to solve this we use odinal encoding which do ranking.
```

```python
from sklearn.preprocessing import OrdinalEncoder
df=pd.DataFrame({
    'size':['small','medium','largest','medium','largest','largest']
})

encoder=OrdinalEncoder(categories=[['small','medium','largest']])
```

```python
encoder.fit_transform(df[['size']])
```

```python
#so now we have ranked the most repeated value so that ml model will understand which is the largest correctly. 
```

```python
#now we do target guided ordinal encoding
# create a sample dataframe with a categorical variable and a target variable
df = pd.DataFrame({
    'city': ['New York', 'London', 'Paris', 'Tokyo', 'New York', 'Paris'],
    'price': [200, 150, 300, 250, 180, 320]  # it is a target variable as it defines the rate of each country.
})
```

```python
mean=df.groupby('city')['price'].mean().to_dict()
```

```python
import pandas as pd
df['city_encoded']=df['city'].map(mean)
```
