```python
# so we handle outliers by using 5 number summnary
# minimum, q1, median, q3, maximum
```

```python
import numpy as np
lst_marks=[-10,30,-70,20,45,39,56,78,98,54,88,76,55,75,20,65,92,100,150,200]
min,q1,median,q3,max=np.quantile(lst_marks,[0,0.25,0.50,0.75,1.0])
```

```python
min,q1,median,q3,max
```

```python
iqr=q3-q1
lowerbound=q1-1.5*(iqr)
upperbound=q3+1.5*(iqr)
print(iqr,'\n',lowerbound,'\n',upperbound)
```

```python
import seaborn as sns
sns.boxplot(lst_marks)
```
